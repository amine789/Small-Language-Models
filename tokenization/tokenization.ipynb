# Building a Simple Tokenizer from Scratch

A step-by-step walkthrough of preprocessing text and building a word-level tokenizer with `encode`/`decode` methods.

## 1. Download the Dataset

Fetch the sample text file (`the-verdict.txt`) used to build our vocabulary.

```python
import os
import requests
import re 
if not os.path.exists("the-verdict.txt"):
    url = (
        "https://raw.githubusercontent.com/rasbt/"
        "LLMs-from-scratch/main/ch02/01_main-chapter-code/"
        "the-verdict.txt"
    )
    file_path = "the-verdict.txt"

    response = requests.get(url, timeout=30)
    response.raise_for_status()
    with open(file_path, "wb") as f:
        f.write(response.content)
```

## 2. Read and Preprocess Text

Load the raw text, split it into tokens (words and punctuation) using a regex, and preview the results.

```python
with open("the-verdict.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
    
print("Total number of character:", len(raw_text))
print(raw_text[:99])
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
all_words = sorted(set(preprocessed))
print(preprocessed[:30])
```

## 3. Build the Vocabulary

Map each unique token to a unique integer ID.

```python


vocab = {token:integer for integer,token in enumerate(all_words)}
```

## 4. Tokenizer Class

Define `SimpleTokenizerV1`, which uses the vocabulary to convert text into token IDs (`encode`) and back into text (`decode`).

```python
class SimpleTokenizerV1:
    def __init__(self, vocab):
        self.str_to_int = vocab
        self.int_to_str = {i:s for s,i in vocab.items()}
    
    def encode(self, text):
        preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', text)
                                
        preprocessed = [
            item.strip() for item in preprocessed if item.strip()
        ]
        ids = [self.str_to_int[s] for s in preprocessed]
        return ids
        
    def decode(self, ids):
        text = " ".join([self.int_to_str[i] for i in ids])
        # Replace spaces before the specified punctuations
        text = re.sub(r'\s+([,.?!"()\'])', r'\1', text)
        return text
```

## 5. Encoding Example

Test `encode()` on a sample sentence to see the resulting list of token IDs.

```python
tokenizer = SimpleTokenizerV1(vocab)
text = """"It's the last he painted, you know," 
           Mrs. Gisburn said with pardonable pride."""
ids = tokenizer.encode(text)
print(ids)

```

## 6. Decoding Example

Test `decode()` to reconstruct the original text from the token IDs.

```python
tokenizer.decode(ids)
```
